7. Dobro

```python
def dobro(numero):
    resultado = numero * 2
    print(f"Número: {numero}")
    print(f"Dobro: {resultado}")

dobro(5)
```

9. Positivo ou negativo

```python
def verificar_numero(numero):
    if numero > 0:
        print("O número é positivo.")
    elif numero < 0:
        print("O número é negativo.")
    else:
        print("O número é zero.")

verificar_numero(4)
verificar_numero(-19)
verificar_numero(0)
```

10. Média

```python
def calcular_media(nota1, nota2):
    media = (nota1 + nota2) / 2
    print(f"Nota 1: {nota1}")
    print(f"Nota 2: {nota2}")
    print(f"Média: {media}")

calcular_media(9, 7)
```
